```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
```

```python
from langchain_groq import ChatGroq

llamaChatModel = ChatGroq(
    model="llama3-70b-8192",  # or "llama-3-8b-8192"
)
```

```python
message = [
    ("system", "You are an marketing expert."),
    ("human", "What is the new idea for facebook markting?"),
]
```

```python
llamaResponse = llamaChatModel.invoke(message)
print(llamaResponse.content)
```
